```python
import os
import yaml
from ultralytics.data.converter import convert_coco
```

```python
# --- 1. Define Paths ---
# You can change this to 'lighting', 'texture', etc., when you process other variants
variant_name = "baseline" 
base_dir = os.path.join("../data/yolo_dataset", variant_name)
annotations_file = os.path.join(base_dir, "coco_annotations.json")
annotations_file, base_dir
```

```python
# --- 2. Convert COCO JSON to YOLO .txt ---
print(f"Converting COCO annotations for variant: {variant_name}...")
convert_coco(
    labels_dir=base_dir,
    # save_dir=base_dir,
    use_segments=True,
    use_keypoints=False,
    cls91to80=False
)
print("Conversion complete! YOLO.txt files are ready.")
```

```python

# --- 3. Build the YOLO data.yaml File ---
# These are the 8 target classes we explicitly isolated earlier
ycbv_classes = [
    "002_master_chef_can",
    "003_cracker_box",
    "005_tomato_soup_can",
    "006_mustard_bottle",
    "011_banana",
    "021_bleach_cleanser",
    "035_power_drill",
    "037_scissors"
]

# YOLO needs to know where the images/labels are, and what the classes are
yaml_data = {
    "path": base_dir,
    "train": "images", # Assuming images are inside coco_data or adjust as needed
    "val": "images",   # We use the same for validation during this synthetic phase
    "labels": "labels",
    "nc": len(ycbv_classes),
    "names": ycbv_classes
}

yaml_path = os.path.join(base_dir, "dataset.yaml")
with open(yaml_path, 'w') as outfile:
    yaml.dump(yaml_data, outfile, default_flow_style=False)

print(f"Generated YOLO config file at: {yaml_path}")
```
